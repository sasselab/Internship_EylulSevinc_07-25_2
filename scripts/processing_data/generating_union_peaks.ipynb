```python
#doesnt work

import os
import numpy as np


#Calculate the genome size

genome_size = 0

chrom_lengths = {}
chrom_order = []
offsets = {}

with open("data/dm6.fa.fai") as f:
    for line in f:
        parts = line.strip().split('\t')
        name = parts[0]
        length = int(parts[1])

        chrom_lengths[name] = length
        chrom_order.append(name)
        offsets[name] = genome_size

        genome_size += length
        

print("Total genome size (including chrUn):", genome_size)


print(offsets)
print(chrom_lengths)
print(chrom_order)
```

```python
#Create a union set array of all peaks

genome_array = np.zeros(genome_size, dtype=np.uint8)

peak_dir = "ATAC-seq_narrowPeak_files"
n_total = 0
n_used = 0

for filename in os.listdir(peak_dir):
    if filename.endswith(".narrowPeak"):
        print(f"\n📄 Processing: {filename}")
        with open(os.path.join(peak_dir, filename)) as f:
            for line in f:
                parts = line.strip().split('\t')
                if len(parts) < 3:
                    continue  # skip malformed lines

                chrom = parts[0]
                start = int(parts[1])
                end = int(parts[2])
                n_total += 1

                if chrom not in offsets:
                    print(f"⚠️ Skipping unknown chrom: {chrom}")
                    continue

                offset = offsets[chrom]
                genome_array[offset + start : offset + end] = 1
                n_used += 1

print(f"\nFinished processing peaks.")
print(f"Total peaks: {n_total}")
print(f"Used peaks: {n_used}")
print(f"Skipped peaks: {n_total - n_used}")
```

```python
#Generate union atac-seq peaks file in narrowPeak format

with open("atac_union.narrowPeak", "w") as out:
    peak_id = 0
    for chrom in chrom_order:
        start_idx = offsets[chrom]
        end_idx = start_idx + chrom_lengths[chrom]
        chrom_array = genome_array[start_idx:end_idx]

        diffs = chrom_array[:-1] - chrom_array[1:]
        starts = np.where(diffs == -1)[0] + 1
        ends = np.where(diffs == 1)[0] + 1

        if chrom_array[0] == 1:
            starts = np.insert(starts, 0, 0)
        if chrom_array[-1] == 1:
            ends = np.append(ends, chrom_array.size)

        for s, e in zip(starts, ends):
            score = 1000
            strand = "."
            signal = 10.0
            pval = 1.0
            qval = 1.0
            summit = (e - s) // 2  # fake summit at center
            out.write(f"{chrom}\t{s}\t{e}\tpeak_{peak_id}\t{score}\t{strand}\t{signal}\t{pval}\t{qval}\t{summit}\n")
            peak_id += 1

print(f"Wrote atac_union.narrowPeak with {peak_id} peaks.")
```
